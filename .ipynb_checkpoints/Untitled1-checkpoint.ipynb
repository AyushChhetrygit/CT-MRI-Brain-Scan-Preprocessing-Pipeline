```python
import numpy as np
import matplotlib.pyplot as plt
import pydicom


# path to your DICOM file
dicom_path = "/Users/ayushchhetry/spartis assignment/JUH_MR-CT_dataset/CT/image_CT/1_CT_s1.dcm"

# read file
ds = pydicom.dcmread(dicom_path)

# extract pixel array
image = ds.pixel_array

print(image.shape)

plt.imshow(image, cmap='gray')
plt.title("DICOM Image")
plt.axis('off')
plt.show()

image = image.astype(np.float32)

image = (image - np.min(image)) / (np.max(image) - np.min(image))

mask_path = "path/to/mask.dcm"   # or .png if masks are PNG

mask_ds = pydicom.dcmread(mask_path)
mask = mask_ds.pixel_array

# normalize mask if needed
mask = (mask > 0).astype(np.float32)

plt.imshow(image, cmap='gray')
plt.imshow(mask, cmap='jet', alpha=0.4)
plt.title("Image + Mask Overlay")
plt.axis('off')
plt.show()
```
